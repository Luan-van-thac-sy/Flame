```python
import pandas as pd
import dill
from google import genai
import time
from tqdm import tqdm, trange

# Sử dụng model Gemini mới (có thể đổi sang gemini-1.5-flash-latest nếu muốn rẻ hơn)
GEMINI_MODEL = "gemini-2.5-flash"

# Client sẽ được khởi tạo ở cell cấu hình API


def call_api(content: str) -> str:
    """Gọi Gemini API bằng Google GenAI SDK mới.

    :param content: prompt, str
    :return: model response, str
    """
    global client
    if client is None:
        raise RuntimeError("Gemini client is not initialized. Run the configuration cell first.")

    response = client.models.generate_content(
        model=GEMINI_MODEL,
        contents=content,
        # Có thể thêm config nếu muốn cố định temperature:
        # config=genai.types.GenerateContentConfig(temperature=0),
    )
    return response.text


def get_msg(content: str) -> str:
    """Gọi Gemini với retry đơn giản."""
    max_retries = 3
    for attempt in range(max_retries):
        try:
            return call_api(content)
        except Exception as e:
            print(f"Attempt {attempt + 1} failed: {e}")
            if attempt < max_retries - 1:
                time.sleep(2 ** attempt)
            else:
                raise


def simplify_note(note):
    prompt = 'Please summarize specific sections from a patient\'s discharge summary: 1. HISTORY OF PRESENT ILLNESS, 2. PAST MEDICAL HISTORY, 3. ALLERGIES, 4. MEDICATIONS ON ADMISSION 5.DISCHARGE MEDICATIONS. Ignore other details while in hospital and focus only on these sections.\n'\
'output template:\n'\
'HISTORY OF PRESENT ILLNESS:\n'\
'(Language summary as short as possible)\n'\
'PAST MEDICAL HISTORY:\n'\
'(Language summary as short as possible)\n'\
'ALLERGIES:\n'\
'(A series of allergies names, separated by commas, does not require any other information)\n'\
'MEDICATIONS ON ADMISSION:\n'\
'(A series of drug names, separated by commas, remove dosage information. Maybe None.)\n'\
'DISCHARGE MEDICATIONS:\n'\
'(A series of drug names, separated by commas, remove dosage information. Maybe None.)\n'\
'Note:' + note + '\n' + 'Summarize result in five aspects in a concise paragraph without any other words:\n'

    msg = get_msg(prompt)

    return msg
```

```python
import pandas as pd
import dill
import os
from pathlib import Path

# Input folder containing .pkl files
input_folder = 'data/data_process/output/mimic-iii/note'

# Output folder for CSV files
output_folder = '/content/csv'

# Create output directory if it doesn't exist
os.makedirs(output_folder, exist_ok=True)

# Get all .pkl files in the input folder
note_folder = Path(input_folder)

# Loop through all .pkl files
for pkl_file in note_folder.glob('*.pkl'):
    try:
        print(f"Processing: {pkl_file.name}...")
        if not pkl_file.name.startswith('allergies'):
            continue

        # Load the pickle file
        data = dill.load(open(pkl_file, 'rb'))

        # Convert to DataFrame if needed
        if not isinstance(data, pd.DataFrame):
            data = pd.DataFrame(data)

        # Create output CSV filename (same name but with .csv extension)
        csv_filename = pkl_file.stem + '.csv'
        csv_path = os.path.join(output_folder, csv_filename)

        # Export to CSV
        data.to_csv(csv_path, index=False, encoding='utf-8')

        print(f"  ✓ Exported {len(data)} rows to {csv_filename}")

    except Exception as e:
        print(f"  ✗ Error processing {pkl_file.name}: {e}")

print(f"\n✅ All files processed! CSV files saved to: {output_folder}")
```

```python
import pandas as pd
import json
from collections import Counter

# Script để check mismatch giữa training data và evaluation results
def check_evaluation_coverage():
    # 1. Load training data
    data = pd.read_csv('data/data_process/output/mimic-iii/data4LLM_with_note.csv')
    split_point = int(len(data) * 2 / 3)
    data_train = data[:split_point]

    # 2. Load evaluation results
    visit_error = pd.read_csv('evaluate_results/cls/train/visit_error.csv')

    # 3. Load cls_train.json để xem evaluate bao nhiêu visits
    with open('data/cls/cls_train.json', 'r') as f:
        cls_train = json.load(f)

    # Tính số visits được evaluate (mỗi visit tạo ra 151 samples trong cls format)
    num_samples = len(cls_train)
    num_visits_in_cls = num_samples // 151  # Vì mỗi visit × 151 thuốc

    print("=" * 70)
    print("EVALUATION COVERAGE ANALYSIS")
    print("=" * 70)
    print(f"\n1. TRAINING DATA:")
    print(f"   Total training visits: {len(data_train)}")

    print(f"\n2. CLS EVALUATION DATA:")
    print(f"   Total samples in cls_train.json: {num_samples}")
    print(f"   Estimated visits evaluated: {num_visits_in_cls}")
    print(f"   Coverage: {num_visits_in_cls / len(data_train) * 100:.1f}%")

    print(f"\n3. EVALUATION RESULTS:")
    print(f"   Visits in visit_error.csv: {len(visit_error)}")

    # 4. Analyze GT Lengths in training data
    train_gt_lengths = []
    for idx in range(len(data_train)):
        drug_ids = eval(data_train.iloc[idx]['drug_id'])
        train_gt_lengths.append(len(drug_ids))

    train_gt_counter = Counter(train_gt_lengths)

    # 5. Analyze GT Lengths in evaluation results
    eval_gt_lengths = visit_error['GT Len'].astype(int).tolist()
    eval_gt_counter = Counter(eval_gt_lengths)

    print(f"\n4. GT LENGTH DISTRIBUTION:")
    print(f"\n   Training data GT Lengths:")
    print(f"   Range: {min(train_gt_lengths)} - {max(train_gt_lengths)}")
    print(f"   Unique GT Lengths: {len(train_gt_counter)}")
    print(f"   Top 10 most common:")
    for gt_len, count in train_gt_counter.most_common(10):
        print(f"      GT Len {gt_len:2d}: {count:3d} visits")

    print(f"\n   Evaluation data GT Lengths:")
    print(f"   Range: {min(eval_gt_lengths)} - {max(eval_gt_lengths)}")
    print(f"   Unique GT Lengths: {len(eval_gt_counter)}")

    # 6. Find MISSING GT Lengths
    all_train_gt = set(train_gt_lengths)
    all_eval_gt = set(eval_gt_lengths)
    missing_gt = sorted(all_train_gt - all_eval_gt)

    print(f"\n5. MISSING GT LENGTHS (in training but NOT in evaluation):")
    print(f"   Count: {len(missing_gt)}")
    if missing_gt:
        print(f"   Missing GT Lengths: {missing_gt}")
        print(f"\n   Details of missing GT Lengths:")
        for gt_len in missing_gt:
            count = train_gt_counter[gt_len]
            print(f"      GT Len {gt_len:2d}: {count:3d} visits affected")
    else:
        print(f"   ✅ No missing GT Lengths! Full coverage.")

    # 7. Calculate impact
    affected_visits = sum(train_gt_counter[gt_len] for gt_len in missing_gt)
    print(f"\n6. IMPACT ANALYSIS:")
    print(f"   Total visits affected by missing GT Lengths: {affected_visits}")
    print(f"   Percentage of training data: {affected_visits / len(data_train) * 100:.2f}%")

    if affected_visits > 0:
        print(f"\n   ⚠️  When generating {affected_visits} visits, will use nearest GT Length approximation")
        print(f"   💡 Recommendation: {'Low impact, acceptable' if affected_visits / len(data_train) < 0.1 else 'Consider evaluating more visits for better coverage'}")

    print("\n" + "=" * 70)

if __name__ == '__main__':
    check_evaluation_coverage()
```
