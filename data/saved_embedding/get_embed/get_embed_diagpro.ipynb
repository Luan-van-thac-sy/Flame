```python
model_path = '/content/Flame/Trained/MICRON/Epoch_39_JA_0.515_DDI_0.06784.model'  ##

diag_embed_path = '/content/Flame/data/saved_embedding/embed/diag_embed_micron.pkl'
pro_embed_path = '/content/Flame/data/saved_embedding/embed/pro_embed_micron.pkl'
```

```python
import torch
import dill
import os

model = torch.load(model_path)

diag_embed = model['embeddings.0.weight'].to('cpu')
pro_embed = model['embeddings.1.weight'].to('cpu')

diag_embed_dict = dict()
pro_embed_dict = dict()

for idx in range(diag_embed.shape[0]):
    diag_embed_dict[idx] = diag_embed[idx]

for idx in range(pro_embed.shape[0]):
    pro_embed_dict[idx] = pro_embed[idx]

# Create output directory if it doesn't exist
os.makedirs(os.path.dirname(diag_embed_path), exist_ok=True)
os.makedirs(os.path.dirname(pro_embed_path), exist_ok=True)

dill.dump(diag_embed_dict, open(diag_embed_path, 'wb'))
dill.dump(pro_embed_dict, open(pro_embed_path, 'wb'))
```
